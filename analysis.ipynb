Pull the data from excel file

```python
import pandas as pd

df = pd.read_excel(r"credit_fraud.xlsx")

df.head()
```

This tells us full data overview like = datatypes, and column structure (understand & analyze the column structure how many columns are there in my datasets & their datatypes)

```python
df.info()
```

This tells us how many missing values per column (i see the missing values in each columns)

```python
df.isnull().sum()
```

This line fill all missing numeric values in my dataset using the middle (median) value of each column (All missing numeric values are filled, each column gets it's own median values except AGE column because in AGE column there are some dirty data including strings like 37.0_err so i converted into numeric datatype)

```python
df = df.fillna(df.median(numeric_only=True))
```

It checks missing values (NaN) in your dataset ( Crosschecking the data if it cleaned or not )

```python
df.isnull().sum()
```

I tried to convert string data into numeric datatype in AGE column 

```python
df['age'] = df['age'].fillna(df['age'].median())
```

converts string data into numeric data ( basically it removes the messy texts into Nan and then it is converted into float numeric datatypes e.g 37.0_err is now = 37.0)

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df['age'] = df['age'].fillna(df['age'].median())
```

It counts how many missing values (NaN) are in the AGE column

```python
df['age'].isnull().sum()
```

so again i check my whole dataset if there is any error and any missing values is remainig or not (It shows how many missing values (NaN) exist in each column of your dataset)

```python
df.isnull().sum()
```

It saves your cleaned Datainto a CSV file

```python
df.to_csv("clean_credit_fraud.csv", index=False)
```
